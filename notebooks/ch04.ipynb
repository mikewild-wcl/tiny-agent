```python
from tiny_agent import TinyAgent
from tiny_agent.llm import LLM
from tiny_agent.memory import Memory
from tiny_agent.tools import Tools

model = "gemma4:e2b"
llm = LLM(model=model)
memory = Memory()
tools = Tools()

# Tell the agent who I am, then ask it if it knows my name
# Before adding memory, it will fail
agent = TinyAgent(llm, memory, tools)
response = agent.run("Hi, I'm Mike Michaelson")
print(response)

response = agent.run("Do you know my name?")
print(response)
```
