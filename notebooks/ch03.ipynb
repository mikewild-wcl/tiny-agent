```python
import pprint as pp
from tiny_agent.llm import LLM
from tiny_agent.tiny_agent import TinyAgent

model = "gemma3:12b"
llm = LLM(model=model, temperature=1)
```

```python
prompt_example = "I saw 6 flamingos. 2 flew away. I hid behind a tree. How many can I see? Be concise."
prompt_answer = "3"

query = "I saw 9 pengions. 2 slid in the water and disappeared from sight while 4 waddled up from the shore.How many can I see?"

messages = [
    {"role": "user", "content": prompt_example},
    {"role": "assistant", "content": prompt_answer},
    {"role": "user", "content": query}
]

response = llm.generate(messages)
pp.pprint(response)
```

```python
# Use a prompt example with more detailed reasoning
prompt_example = "I saw 6 flamingos. 2 flew away. I hid behind a tree. How many can I see? Be concise."
prompt_answer = "You say 6 flamingos. 2 flew away. 6 - 2 = 4. Then 1 flamingo flew away. You now see 4 - 1 = 3."

messages = [
    {"role": "user", "content": prompt_example},
    {"role": "assistant", "content": prompt_answer},
    {"role": "user", "content": query}
]

response = llm.generate(messages)
pp.pprint(response)
```

```python
# Step-by-step reasoning example
messages = [{"role": "user", "content": query + " Let's thing step by step"}]

response = llm.generate(messages)
pp.pprint(response)
```

```python
agent = TinyAgent(llm)
response = agent.run(query + " Let's think step by step.")
pp.pprint(response)
```
